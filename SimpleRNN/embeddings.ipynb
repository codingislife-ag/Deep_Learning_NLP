```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
## Define the vocab size

voc_size = 10000
```

```python
## One Hot Representation 

one_hot_repr = [one_hot(words, voc_size) for words in sent]
one_hot_repr
```

```python
# word embedding representation

from tensorflow.keras.layers import Embedding
# from tensorflow.keras.processing.sequence import pad_sequences  # deprecated
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
# to make the length of the all the sentences equal

sent_length = 8         # take the max_length across all sentences, or more than that

# either we can add zeros in the beginning of the sentence to make them equal to 8 or at the last(padding = 'post')
embedded_docs = pad_sequences(one_hot_repr, padding = 'pre', maxlen = sent_length)
print(embedded_docs)
```

```python
## feature representation

dim = 10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_shape=(sent_length,)))
model.compile('adam', 'mse')        # adam optimiser, mse loss function
```
