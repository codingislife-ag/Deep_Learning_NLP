```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
## Define the vocab size

voc_size = 10000
```

```python
i = 1
for word in sent:
    print(word)
    print(i)
    i += 1
```

```python
## One Hot Representation 

one_hot_repr = [one_hot(words, voc_size) for words in sent]
one_hot_repr
```

```python
print(type(one_hot_repr))
```

```python
# word embedding representation

from tensorflow.keras.layers import Embedding
# from tensorflow.keras.processing.sequence import pad_sequences  # deprecated
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
# to make the length of the all the sentences equal

sent_length = 8         # take the max_length across all sentences, or more than that

# either we can add zeros in the beginning of the sentence to make them equal to 8 or at the last(padding = 'post')
embedded_docs = pad_sequences(one_hot_repr, padding = 'pre', maxlen = sent_length)
print(embedded_docs)
```

```python
## feature representation

dim = 10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_shape=(sent_length,)))     # just like we add hidden layers by using model.add(Dense())
model.compile('adam', 'mse')        # adam optimiser, mse loss function
```

```python
model.summary()
```

```python
# every word is represented by 10 dimensions here

model.predict(embedded_docs)
```

```python
'''

🔴 Problem
model.predict(embedded_docs[0])

This throws:

ValueError: Cannot take the length of shape with unknown rank
🔍 Root Cause (VERY IMPORTANT)

Your model expects input shape:

(input_shape=(8,))

So Keras expects:

(batch_size, 8)
✅ What you're passing
embedded_docs[0]

Shape:

(8,)

👉 This is 1D (no batch dimension)

❌ What model expects
(1, 8)

👉 2D input (batch included)

🧠 Key Concept
Keras ALWAYS expects batch dimension

Even for a single sample:

Case	Shape
Single sentence	(1, 8)
Multiple sentences	(n, 8)
✅ Fix (Simple)
✔ Option 1: Use reshape
model.predict(embedded_docs[0].reshape(1, -1))
✔ Option 2: Use slicing (cleaner)
model.predict(embedded_docs[0:1])

👉 This keeps it 2D automatically

✔ Option 3: Use np.expand_dims
np.expand_dims(embedded_docs[0], axis=0)
🔬 Why This Error Happens (Advanced Insight)

When you pass:

(8,)

Keras cannot determine:

Is this 1 sample with 8 features?

Or 8 samples with 1 feature?

👉 Ambiguity → leads to "unknown rank" error

🔥 What Your Model Is Actually Doing

Let’s connect everything:

Input
(1, 8)
After Embedding
(1, 8, 10)

Meaning:

1 sentence

8 words

each word → 10-dimensional vector

🧠 Bonus Insight (VERY IMPORTANT FOR INTERVIEWS)
Why embedding works here:

Your pipeline:

Text → one_hot → integers → embedding → dense vectors

Example:

"glass" → 345 → [0.2, -0.1, ..., 0.7]

👉 Embedding layer = trainable lookup table

⚡ Extra Tip (Subtle but Important)

This line:

model.compile('adam', 'mse')

👉 You're not actually training anything (no fit())

So:

embeddings are random

predict() just returns random vectors

✅ Final Fix Summary

Replace:

model.predict(embedded_docs[0])

With:

model.predict(embedded_docs[0:1])
'''
```
