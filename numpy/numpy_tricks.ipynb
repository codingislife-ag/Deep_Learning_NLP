#### np.sort

```python
# python sort always returns a list, thats why we are using np.sort
# np.sort returns numpy array in sorted form
import numpy as np
a = np.random.randint(1, 100, 15)
print(a)
np.sort(a)
```

```python
# Sort in descending Order
np.sort(a)[::-1]
```

```python
b = np.random.randint(1, 100, 24).reshape(6, 4)
print(b)
print(np.sort(b))
```

```python
# sort column wise
np.sort(b, axis=0)
```

### 🔑 The Golden Rule (MOST IMPORTANT)

👉 axis tells you the direction you move ALONG, not what you operate ON

axis=0 → move down the rows (vertical direction ⬇️)
axis=1 → move across columns (horizontal direction ➡️)

### np.concatenate

np.concatenate() function concatenate a sequence of arrays along an existing axis.

```python
c = np.arange(6).reshape(2, 3)
d = np.arange(6, 12).reshape(2, 3)

print(c)
print(d)
```

```python
np.concatenate((c, d), axis=0)
```

```python
np.concatenate((c, d), axis=1)
```

### np.where

The numpy.where() function returns the indices of elements in an input array where the given array condition is satisfied.

```python
a
```

```python
# find all indices with value > 50
np.where(a > 50)
```

```python
# replace all values > 50 with 0
np.where(a > 50, 0, a)
```

```python
# replace all even numbers by 0
np.where(a%2==0, 0, a)
```

### np.percentile

```python
# 0 percentile will always be minimum, because no 
np.percentile(a, 0)
```
