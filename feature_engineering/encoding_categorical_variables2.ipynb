```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.base import BaseEstimator, TransformerMixin
import pandas as pd
import sklearn
```

```python
class CountEncoder(BaseEstimator, TransformerMixin):
    def __init__(self, columns=None):
        self.columns = columns
        self.count_map = {}

    def fit(self, X, y=None):
        if self.columns is None:
            self.columns = X.columns
        for col in self.columns:
            self.count_map[col] = X[col].value_counts().to_dict()
        return self
    
    def transform(self, X):
        X = X.copy()
        for col in self.columns:
            X[col] = X[col].map(self.count_map[col]).fillna(0)
        return X
```

```python
preprocessor = ColumnTransformer(
    transformers=([
        ('age_missing', SimpleImputer(strategy='mean'), ['Age']),
        ('cat_state', CountEncoder(), ['State']),
        ('education_ordinal', OrdinalEncoder(), ['Education'])
    ])
)

sklearn.set_config(transform_output='pandas')   # just so we can get the output in pandas
```

```python
preprocessor.fit_transform(X_train)
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

state_choices = np.array(
    ['Karnataka', 'Tamil Nadu', 'Maharashtra', 'Delhi', 'Telangana', np.nan],
    dtype=object
)

education_choices = np.array(
    ['High School', 'UG', 'PG', np.nan],
    dtype=object
)

data = {
    'State': np.random.choice(state_choices, size=100),
    'Education': np.random.choice(education_choices, size=100)
}

df = pd.DataFrame(data)
```

```python
df.isnull().sum()
```

```python
# Initialize the CountEncoder with various parameters

encoder = ce.CountEncoder(
    cols=['State', 'Education'],
    handle_missing='value',     # Treat NaNs as a separate/countable category
    handle_unknown='error',     #Treat unknown categories as NaNs (if seen during transformation not fit)
)
```

```python
# Fit and transform the dataset

encoded_df = encoder.fit_transform(df)
print(encoded_df.head(5))
```

```python
'''
We have 3 options for
handle_missing = 
1. 'value' -> treat np.nan/missing values at separate category.
2. 'return nan' -> dont encode the missing values, just return them as it is.
3. 'error' -> if our dataframe has missing values, it wil get errored out. 


Handle_unknown -> lets say, we trained/ fitted the data on 5 different categories but at the time of 
transformation, we get some unseen value which has never been seen before during fitting/ training.
here we also have multiple option:-

1. 'error' -> throw error, simply.
2. 'return_nan' -> return nan instead of those categories
3. 'value' -> assign them new category, most probably, it will return zero because their frequency will 
obviously be zero.
'''
```
