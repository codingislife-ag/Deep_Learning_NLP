### Attribute	Meaning
1. categories	-> What you provided manually
2. categories_	-> What the model learned during fit()

One more subtle ML detail

When you manually define categories, OrdinalEncoder will NOT learn new ones.

Example:

If test data contains "Excellent" → it will throw an error.

To handle this we often use:

OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)

#### how to handle unknown values

```python
oe = OrdinalEncoder(
    categories=[['Poor', 'Average', 'Excellent'], ['School', 'UG', 'PG']],
    handle_unknown='use_encoded_value',
    unknown_value=-1,
)

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :2], df.iloc[:, -1], test_size=0.2, 
                                                    random_state=42)
X_train = oe.fit_transform(X_train)
oe.transform(np.array(['Poor', 'College']).reshape(1, 2))
```

#### Handling infrequent (Rare) Categories

```python
X = np.array([['dog'] * 5 + ['cat'] * 20 + ['rabbit'] * 10 + ['snake'] * 3 + ['horse'] * 2], dtype=object).T
X
```

```python
'''it sorts the categories by their frequencies and then picks the (max_categories-1) from the top
 and then combines the remaining which is going to the last category.
'''

enc = OrdinalEncoder(max_categories=3).fit(X)
```

```python
enc.infrequent_categories_
```

```python
enc.transform(np.array([['cat', 'rabbit', 'snake', 'dog', 'horse']]).reshape(5, 1))
```

```python
''' 
only consider the categories which have frequency >= min_frequency
rest of them -> consider in others
'''

enc = OrdinalEncoder(min_frequency=4).fit(X)
enc.infrequent_categories_
```

```python
enc.transform(np.array([['cat', 'rabbit', 'snake', 'dog', 'horse']]).reshape(5, 1))
```

#### Deal with Missing Values

```python
# categorical data with missing values
data = [['Cat'], [np.nan], ['Dog'], ['Fish'], [np.nan]]

# setting encoded_missing_values to -1, indicating we want missing values to be encoded as -1
encoder = OrdinalEncoder(encoded_missing_value=-1)
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```

```python
# or set it to np.nan
encoder = OrdinalEncoder(encoded_missing_value=np.nan)
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```
