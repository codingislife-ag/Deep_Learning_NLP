# The Easy Way -> Using Column Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('cars.csv')
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=['selling_price']),
    df['selling_price'],
    test_size=0.2,
    random_state=42
)
```

```python
X_train.head()
```

```python
transformer = ColumnTransformer(
    [
        ("ordinal", OrdinalEncoder(categories=[['Test Drive Car', 'Fourth & Above Owner', 'Third Owner', 'Second Owner', 'First Owner']]), ['owner']),
        ("onehot", OneHotEncoder(sparse_output=False), ['brand', 'fuel'])
    ],
    remainder='passthrough'
)

# setting to get pandas dataframe, without this, we were getting numpy array earlier.
transformer.set_output(transform='pandas')

'''
ColumnTransformer(
    [("operation_name", object, [col1, col2, .....]), (), ()]
)

operation_name -> whatevr we want to give, our choice. whatever we want to call that operation/transformation
'''
```

```python
# giving full data, X_train and X_test

X_train_transformed = transformer.fit_transform(X_train)
X_test_transformer = transformer.transform(X_test)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import SelectKBest, chi2

from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

```python
'''
ColumnTransformer([
    (name, transformer, columns)
], remainder=...)


Your case
('scale', MinMaxScaler(), slice(0, 38))

Meaning:

Element	Meaning
scale	        Name of transformation
MinMaxScaler()	Scaling algorithm
slice(0,38)	    Columns 0 → 37

So:
Apply MinMaxScaler to columns 0–37
'''
```

```python
# imputation transformer

trf1 = ColumnTransformer([
    ('impute_km_driven', SimpleImputer(), [1]),   # default setting -> it replaces the missing value with the mean
    ('impute_owner', SimpleImputer(strategy='most_frequent'), [3]),  # here it replaces with the most frequent/common values -> in our case, it is 'First Owner'
], remainder='passthrough')
```

```python
# encoding categorical variables

trf2 = ColumnTransformer([
    ('ordinal', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), [1]),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False), [2, 3]),
], remainder='passthrough')
```

```python
# Scaling
# at this stage, we have 38 cols in total -> [0->37]
# also learn about Standard Scaler

trf3 = ColumnTransformer([
    ('scale', MinMaxScaler(), slice(0, 38)),
], remainder='passthrough')
```

```python
a = [1, 2, 3, 4, 5]
x = slice(0, 4)     # this is a slice object.
print(x)  # we can also try printing -> it is a slice object -> slice(0, 5, None)

'''
start = 0
stop = 4
step = None

The slice object simply stores slicing parameters.

You can check its attributes:

print(x.start)  # 0
print(x.stop)   # 4
print(x.step)   # None

3. Why does slice() exist?

Because sometimes the slice must be stored or passed dynamically.

Example:

def get_part(lst, sl):
    return lst[sl]

a = [1,2,3,4,5]

s = slice(1,4)

print(get_part(a, s))

5. Does a[x] modify the list in-place?

No.

a = [1,2,3,4,5]
x = slice(0,4)

b = a[x]

print(a)
print(b)

Output:

[1,2,3,4,5]
[1,2,3,4]

a[x] returns a new list.

It does NOT modify a in place.

'''


b = a[x]  # this is equivalent to a[0:4]
print(a)
print(b)
```

```python
'''
Also, why are we using slice??

2️⃣ What ColumnTransformer Accepts for Column Selection

ColumnTransformer allows multiple ways to specify columns:

Method	Example
List of column indices	[0,1,2]
Single column	[3]
Slice object	slice(0,38)
Column names	['age','salary']
Boolean mask	[True,False,True]

So sklearn allows flexibility.

3️⃣ Why slice(0,38) Is Used Here

At this stage you said:
we have 38 columns -> 0 to 37
Instead of writing:
[0,1,2,3,4,5,6,7....37]
we simply write:
slice(0,38)
Which means:
columns 0 → 37
It is shorter and cleaner.

4️⃣ Could We Use List Instead?
Yes. This would work exactly the same:
trf3 = ColumnTransformer([
    ('scale', MinMaxScaler(), list(range(38))),
], remainder='passthrough')

Because:
list(range(38))
produces
[0,1,2,3,4,5,6,...,37]
But this is less elegant.


9️⃣ Important Subtle Point (Very Useful)
If your data is a pandas DataFrame, you could also do:
('scale', MinMaxScaler(), X.columns)
Which means:
scale all columns
Even better.

'''
```

```python
'''
### Your current approach works, but it has three major problems:

1️⃣ Column indices keep changing
2️⃣ Hardcoding slice(0,38) is fragile
3️⃣ Multiple ColumnTransformers make debugging harder

The better design is:

ONE ColumnTransformer with small pipelines inside it

This avoids index shifting completely.

⭐ Professional Pipeline Structure

Instead of:

Imputer → Encoder → Scaler → FeatureSelector → Model

we build column-wise pipelines.

                ColumnTransformer
               /        |        \
        numeric      owner      categorical
         pipeline    pipeline      pipeline
            |           |              |
       impute+scale  impute+ord    onehot

Then attach model.

Step 1 — Define column groups

Since your dataset is:

column	type
company	categorical
km_driven	numeric
fuel	categorical
owner	ordinal categorical
num_cols = [1]        # km_driven
owner_col = [3]       # owner
cat_cols = [0,2]      # company, fuel
Step 2 — Create small pipelines
Numeric pipeline
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', MinMaxScaler())
])
Owner pipeline (ordinal)
owner_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder())
])
Categorical pipeline
cat_pipeline = Pipeline([
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
Step 3 — Combine them
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('owner', owner_pipeline, owner_col),
    ('cat', cat_pipeline, cat_cols)
])

Now sklearn automatically handles:

imputation

encoding

scaling

without column index shifting.

Step 4 — Full pipeline
pipe = Pipeline([
    ('preprocess', preprocessor),
    ('feature_select', SelectKBest(chi2, k=10)),
    ('model', RandomForestRegressor())
])

Then:

pipe.fit(X_train, y_train)
⭐ Why this approach is better
Your Approach	Professional Approach
Multiple ColumnTransformers	Single ColumnTransformer
Indices change after each step	Indices never change
Hardcoded slice(0,38)	No guessing columns
Harder debugging	Very clean structure
⭐ Even Better (automatic column detection)

If your dataset grows, you can automatically detect column types.

num_cols = X.select_dtypes(include=['int64','float64']).columns
cat_cols = X.select_dtypes(include=['object']).columns

This is very common in real ML projects.

⭐ Bonus: Debugging trick

You can inspect the transformed dataset:

pipe.named_steps['preprocess'].fit_transform(X_train)

or feature names:

pipe.named_steps['preprocess'].get_feature_names_out()
⭐ Important ML Interview Concept

This pipeline automatically prevents data leakage.

Without pipelines people accidentally do:

imputer.fit(X)
scaler.fit(X)
train_test_split()

Which leaks test data statistics into training.

Pipelines ensure:

fit only on training data
apply to test data

This is one of the main reasons sklearn pipelines exist.

✅ Since you're learning ML pipelines deeply, the next very powerful concept you should know is:

'''
```

```python
'''
8️⃣ Another important problem you will soon hit

You used:

SelectKBest(score_func=chi2)

But chi² only works with non-negative values.

So if you used:

StandardScaler

(which produces negative values)

you would get another error.

That's why:

chi2 + MinMaxScaler works
chi2 + StandardScaler fails

9️⃣ A debugging trick (very useful)

Whenever pipelines break, print intermediate output:

pipe[:-1].fit_transform(X_train)

or step-by-step:

X1 = trf1.fit_transform(X_train)
X2 = trf2.fit_transform(X1)

This helps see column order changes.
'''
```
