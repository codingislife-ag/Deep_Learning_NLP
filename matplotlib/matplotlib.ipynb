```python
# if I want to see the points

plt.plot(batsman['index'], batsman['V Kohli'], color='#ED42E3', linestyle='solid', linewidth=3, marker='D')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#F7E72F', linestyle='dashdot', linewidth=2, marker='+')
# we have many types of markers, D + > . o

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# we can also control the marker size
plt.plot(batsman['index'], batsman['V Kohli'], color='#ED42E3', linestyle='solid', linewidth=3, marker='D', markersize=10)
plt.plot(batsman['index'], batsman['RG Sharma'], color='#F7E72F', linestyle='dashdot', linewidth=2, marker='+')
# we have many types of markers, D + > . o

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# we can also give names to each and every line
plt.plot(batsman['index'], batsman['V Kohli'], color='#ED42E3', linestyle='solid', 
         linewidth=3, marker='D', markersize=10, label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#F7E72F', linestyle='dashdot', linewidth=2, 
         marker='+', label='Rohit')

# we have many types of markers, D + > . o
plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

# without this, the index for the lines will not show up
plt.legend(loc='best')      # this one is default, the best position.
# plt.legend(loc='right') # we have right, upper right and many more.
```

```python
# size

tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size'])
```

```python
# since the number of family members with every person who paid the bill are mostly very less
# thats why mutliplying it by 20 just to make the size of the dots bigger.

plt.scatter(tips['total_bill'], tips['tip'], s=tips['size'] * 20)

# now you will be able to see the diff between the size of the dots, different size of dots
# for the different number of persons he came with.

# so now this graph will be able to represent 3 information, it is representing information for 3 dimensions
# bigger dots -> for more customers
# smaller dots -> for less customers
```

```python
# Scatterplot using plt.plot

plt.plot(tips['total_bill'], tips['tip'])
```

### plt.plot() -> faster
### plt.scatter() -> slower

```python
# Scatterplot using plt.plot
# this is a faster technique
# but some of the functionalities might not work like dot size and rendering the proper color.
# for the scatter graphs.
plt.plot(tips['total_bill'], tips['tip'], 'o')
```

### 3rd Type of Graphs -> Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use Case - Aggregate analysis of groups

```python
# simple bar chart

children = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, children)
```

```python
# same thing here also, we can change the color
# add the y-label and x-label and title

plt.bar(colors, children, color='red')
plt.title('Colors Used by Children Quantitative Analysis')
plt.xlabel('Colors')
plt.ylabel('Children')

# so we can do almost all of the things we have done in the past with the above 2 charts.
```

```python
# horizontal bar chart

plt.barh(colors, children, color='red')
plt.title('Colors Used by Children Quantitative Analysis')
plt.xlabel('Colors')
plt.ylabel('Children')
```

### Stacked Bar Chart

```python
plt.bar(df['batsman'], df['2015'], label='2015')
plt.bar(df['batsman'], df['2016'], bottom=df['2015'], label='2016')
plt.bar(df['batsman'], df['2017'], bottom=df['2016'] + df['2015'], label='2017')

plt.legend()
plt.show()
```
